```python
import pandas as pd
import numpy as np
from faker import Faker
import random

# Inicializar Faker
fake = Faker()

# Parámetros del dataset
num_rows = 1000  # Número de registros
regions = ["North", "South", "East", "West"]
products = ["Product_A", "Product_B", "Product_C", "Product_D"]

# Generar datos sintéticos
data = {
    "Region": [random.choice(regions) for _ in range(num_rows)],
    "Product": [random.choice(products) for _ in range(num_rows)],
    "Date": [fake.date_between(start_date="-1y", end_date="today") for _ in range(num_rows)],
    "Sales": [random.randint(500, 5000) for _ in range(num_rows)]
}

# Crear DataFrame
df = pd.DataFrame(data)

# Guardar en un archivo CSV
csv_file = "synthetic_sales_data.csv"
df.to_csv(csv_file, index=False)

print(f"Archivo guardado como {csv_file}")
df.head()  # Mostrar las primeras filas
```
